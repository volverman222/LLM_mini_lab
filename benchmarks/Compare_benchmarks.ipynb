```python
import matplotlib.pyplot as plt
import json
```

```python
with open("benchmarks_result_Arc_easy.txt", "r", encoding="utf-8") as file:
    data = json.load(file)
with open("benchmarks_result_Hella_Swag.txt", "r", encoding="utf-8") as file:
    data2 = json.load(file)
```

```python
import matplotlib.pyplot as plt

nombres = ["ARC-Easy", "HellaSwag"]
precisiones = [data["acc"], data2["acc"]]

plt.bar(nombres, precisiones, color=["steelblue", "darkorange"])

plt.xlabel("Benchmark")
plt.ylabel("Precisión")
plt.title("Benchmarks GPT 2 124 M OpenAI")
plt.ylim(0, 1)  # útil si la precisión está entre 0 y 1

for i, precision in enumerate(precisiones):
    plt.text(i, precision + 0.02, f"{precision:.3f}", ha="center")

plt.show()
```
